# Dataset Conversion - Native Ads Detection

**Penelitian Postdoc**: Pengembangan Agentic AI untuk Deteksi Native Ads

Notebook ini mengkonversi dataset Excel (12,088 samples) ke format LLM.

```python
import pandas as pd
import json
from pathlib import Path
import random

random.seed(42)
```

## Load Dataset

```python
# Load Excel
df = pd.read_excel("../native_ads_dataset.xlsx", sheet_name='Clean')
print(f"✅ Loaded {len(df)} samples")
df.head()
```

## Convert to Instruction Format

```python
instruction_data = []

for idx, row in df.iterrows():
    content = str(row['content'])
    label = str(row['label'])
    
    instruction_data.append({
        "id": f"sample_{idx}",
        "instruction": "Klasifikasikan artikel berikut sebagai native advertising atau konten editorial.",
        "input": content[:1000],
        "output": f"**Klasifikasi**: {label}"
    })

print(f"✅ Converted {len(instruction_data)} samples")
```

## Save

```python
# Create output directory
output_dir = Path('../data')
output_dir.mkdir(exist_ok=True)

# Save JSON
with open(output_dir / 'llm_dataset_instruction.json', 'w', encoding='utf-8') as f:
    json.dump(instruction_data, f, indent=2, ensure_ascii=False)

print("✅ Saved to: data/llm_dataset_instruction.json")
print(f"\nFile size: {(output_dir / 'llm_dataset_instruction.json').stat().st_size / 1024 / 1024:.2f} MB")
```
